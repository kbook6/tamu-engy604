```python
# processes: PV, WF
# storage: LiI
# resources: wind, solar, power (demand of)
# currency: USD
# periods: time, q (quarters) y (year)
```

```python
from energia import Process, Storage, Resource, Currency, Periods, Model
```

Small Example

```python
m = Model()
m.q = Periods(label = 'Quarter of the Year')
m.y = 4*m.q # a year is made of 4 quarters

# side note: selecting for specific quarters:
# quarters are indexed at 0
# 0 = first quarter
#m.q[2] # gives an output
# -----

print('list of periods', m.periods)
print('y to q conversion:', m.y.howmany(m.q))
print('q to y conversion:', m.q.howmany(m.y))
print('temporal horizon:', m.horizon) # horizon should always be 1 year in this case
# -----

m.usd = Currency() # single equals sign for a declaration, double equals for behaviors
m.declare(Resource, ['power','wind','solar'])
print('list of resources:', m.resources)

m.wind.consume == True
m.solar.consume == True
m.power.release.prep(180) >= [0.6, 0.7, 0.8, 0.3] # scaled to nominal value

# wind
m.wf = Process()
m.wf(m.power) == -m.wind
# wf constraints (upper and lower):
m.wf.capacity.x <=300 # arbitrary number
m.wf.capacity.x >=10 # the .x sets it as a binary so if there is no wind farm, it doesn't return an error with >=10
# -----
# capex = capital expenditure, scales with size (capacity) of project
m.wf.capacity[m.usd.spend] == 993991
# can also do m.wf.capex == 993991
# -----
# opex = operation expenditure | mapping constraint
m.wf.operate <= [0.9, 0.8, 0.5, 0.7] # how use per each quarter
m.wf.operate[m.usd.spend] == 49 # essentially forces it to multiply 49 in each operate term for each quarter and then sum it all (line 17)
# can also do m.wf.opex == 49 # can also set the opex as a [] to define each quarter's opex

# solar
m.pv = Process()
m.pv(m.power) == -m.solar
m.pv.capacity.x <= 300
m.pv.capacity.x >= 10
m.pv.capex == 1439046
m.pv.operate <= [0.6, 0.8, 0.9, 0.7]
m.pv.opex == 90000

# battery
m.lii = Storage()
m.lii(m.power) == 0.9
m.lii.capacity.x <= 300
m.lii.capacity.x >= 10
m.lii.capex == 100000
m.lii.charge.capacity <= 100
m.lii.discharge.capacity <= 100

print('wind:')
m.wf.show(True)
print('')
print('solar')
m.pv.show(True)
print('')
print('battery')
m.lii.show(True)

m.locate(m.wf, m.pv, m.lii)
m.usd.spend.opt()

m.sol()
```
