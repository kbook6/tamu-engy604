## Define the Model for Part A

```python
### Setting the Model ###
m = Model()
m.y = Periods(label='Year')
m.operating_horizon = m.y

m.usd = Currency()
```

```python
##### problem notes #####
### model equation:
# c = consumed
# d = discharged
# p = produced (delta)
#
# c - d + p = 0

### resources
# grid electricity, biomass, natural gas, solar, wind; electricity, heaing; refrigeration, lighting, heating

### processes
# biomass ST, natural gas CHP, solar PV, wind farm; refrigerator, LED, space heater
```

```python
### setting up resources ###
# consumed resources: biomass, natural gas, grid electricity, solar electricity, wind electricity
# implicity produced resources: electricity, heat
# unfixed demand resources: co2
# fixed demand resources: lighting, refrigeration, heating
m.declare(Resource, ['bm','ng','elec_g','solar','wind','heat','electricity','co2', 'lighting', 'heating', 'refrigeration'])
print(m.resources)

# resources that are consumed initially
m.bm.consume == True
m.ng.consume == True
m.elec_g.consume == True
m.solar.consume == True
m.wind.consume == True

# implicit resources: heat and electricity
# these resources are produced after the first processes, but then consumed in the second processes

# resources that are produced
m.refrigeration.release == True
m.lighting.release == True
m.heating.release == True
m.co2.release == True
#######################################################################################################
```

```python
### setting up processes ###
# primary processes: biomass ST, natural gas CHP, solar PV, wind farm, grid
# secondary processes: refrigerator, LED, space heater
m.st = Process()
m.chp_heat = Process()
m.chp_elec = Process()
m.pv = Process()
m.wf = Process()
m.grid = Process()
m.rf = Process()
m.led = Process()
m.spaceheater = Process()
```

```python
## PRIMARIES:
# biomass ST
m.st(m.electricity) == 0.68 * -m.bm # bm -> electricity
m.st.capacity.x >= 100
m.st.capacity.x <=1e6
m.st.capex == 250
m.st.opex == 15
m.st.label == 'Biomass ST'
m.st.operate == True

# natural gas CHP - HEAT
m.chp_heat(m.heat) == 0.28 * -m.ng # nat gas -> heat
m.chp_heat.capacity.x >= 800
m.chp_heat.capacity.x <= 1e6
m.chp_heat.capex == 500
m.chp_heat.opex == 15
m.chp_heat.label == 'Natural Gas CHP - Heat'
m.chp_heat.operate == True

# natural gas CHP - ELECTRICITY
m.chp_elec(m.electricity) == 0.44 * -m.ng # nat gas -> electricity
m.chp_elec.capacity.x >= 800
m.chp_elec.capacity.x <= 1e6
m.chp_elec.capex == 500
m.chp_elec.opex == 15
m.chp_elec.label == 'Natural Gas CHP - Electricity'
m.chp_elec.operate == True

# link ng chp capacities
m.chp_capacity_sync = Constraint(expr = m.chp_heat.capacity.x == m.chp_elec.capacity.x)

# solar pv
m.pv(m.electricity) == 0.09 * -m.solar # solar -> electricity
m.pv.capacity.x >= 10
m.pv.capacity.x <= 300
m.pv.capex == 2000
m.pv.opex == 500
m.pv.label == 'Solar PV'
m.pv.operate == True

# wind farm
m.wf(m.electricity) == 0.22 * -m.wind # wind -> electricity
m.wf.capacity.x >= 10
m.wf.capacity.x <= 500
m.wf.capex == 2000
m.wf.opex == 1200
m.wf.label == 'Wind Farm'
m.wf.operate == True

# grid
m.grid(m.electricity) == 1 * -m.elec_g
m.grid.capacity.x >= 0 # set arbitrary lower bound
m.grid.capacity.x <= 1e6 # set arbitrary upper bound
m.grid.label == 'Grid Electricity'
m.grid.operate == True
#######################################################################################################
```

```python
## SECONDARIES
# refrigerator
m.rf(m.refrigeration) == 3.0 * -m.electricity
m.rf.capacity.x >= 0 # set arbitrary lower bound
m.rf.capacity.x <= 1e6 # set arbitrary upper bound
m.rf.capex == 70
m.rf.opex == 4
m.rf.label == 'Refrigeration'
m.rf.operate == True

# LED
m.led(m.lighting) == 0.8 * -m.electricity
m.led.capacity.x >= 0 # set arbitrary lower bound
m.led.capacity.x <= 1e6 # set arbitrary upper bound
m.led.capex == 10
m.led.opex == 2 # for some reason, we were getting an error when we tried using 1 or 1.0
m.led.label == 'LED Lighting'
m.led.operate == True

# space heater
m.spaceheater(m.heating) == 0.85 * -m.heat
m.spaceheater.capacity.x >= 0 # set arbitrary lower bound
m.spaceheater.capacity.x <= 1e6 # set arbitrary upper bound
m.spaceheater.capex == 30
m.spaceheater.opex == 3
m.spaceheater.label == 'Space Heater'
m.spaceheater.operate == True
#######################################################################################################
```

```python
# locate the processes
m.locate(m.st, m.chp_heat, m.chp_elec, m.pv, m.wf, m.grid, m.rf, m.led, m.spaceheater)
#######################################################################################################
```

```python
### Cost of Resources ###
# price for grid electricity resource
# 1 GJ = 277.78 kWh, so $36.11/GJ = 36.11 / 277.78 = ~0.13 $/kWh
# 0.13 $/kWh * 8760 hr/yr
price_elec_g = (36.11/277.78) * 8760 # price per kilowatt hour for the entire year
print("price of grid electricity: ", price_elec_g)
m.elec_g.price.prep(price_elec_g)

# price for natural gas
price_natural_gas = (8.89/277.78) * 8760
print("price of natural gas: ", price_natural_gas)
m.ng.price.prep(price_natural_gas)

# price for biomass
price_bm = (9.72/277.78) * 8760
print("price of biomass: ", price_bm)
m.bm.price.prep(price_bm)

# wind and solar are free
m.wind.price.prep(0)
m.solar.price.prep(0)
#######################################################################################################
```

```python
### printing price conversions ###
print(8.89/277.78)
print(9.72/277.78)
print(36.11/277.78)
```

```python
### Resource Emissions ###
# set co2 generation
m.gwp = Environ()

# 1 PJ = 2.7778e8 kWh
pj_to_kwh = 2.7778e8
co2_ng = (56/pj_to_kwh)    # kg CO2 per kWy for natural gas
co2_bm = (100/pj_to_kwh)   # kg CO2 per kWy for biomass
co2_elec = (90/pj_to_kwh)  # kg CO2 per kWy for grid electricity

m.gwp.emit(m.chp_heat.capacity) == co2_ng
m.gwp.emit(m.chp_elec.capacity) == co2_ng
m.gwp.emit(m.st.capacity) == co2_bm
m.gwp.emit(m.grid.capacity) == co2_elec

print('the total possible emissions of each CO2 producing technology based on their capacities')
print('natural gas annual co2 emissions:', co2_ng,'kg/kWy')
print('biomass annual co2 emissions:', co2_bm,'kg/kWy')
print('grid electricity co2 emissions:', co2_elec,'kg/kWy')
#######################################################################################################
```

```python
### printing emissions conversions ###
print(56/2.7778e8)
print(100/2.7778e8)
print(90/2.7778e8)
```

### Q1: Operating Horizon

```python
### ### A1 ### ###
### final demand ###

m.lighting.release = True
m.refrigeration.release = True
m.heating.release = True

m.lighting_min_production = Constraint(expr = m.lighting.produce >= 200)
m.refrigeration_min_production = Constraint(expr = m.refrigeration.produce >= 1000)
m.heating_min_production = Constraint(expr = m.heating.produce >= 100) # space heating

# print the process constraints
print('biomass st')
m.st.show(True)
print('')
print('natural gas chp - heat')
m.chp_heat.show(True)
print('')
print('natural gas chp - elec')
m.chp_elec.show(True)
print('')
print('solar pv')
m.pv.show(True)
print('')
print('wind farm')
m.wf.show(True)
print('')
print('grid')
m.grid.show(True)

#######################################################################################################
```

### Q2: Integer Cuts

```python
### ### A2 ### ###
### integer constraints ###
m.integer_cut = m.st.capacity.x + m.chp_elec.capacity.x + m.pv.capacity.x + m.wf.capacity.x + m.grid.capacity.x <= 1
```

### Q3: Minimize Total Cost

```python
### ### A3 ### ###
### min total sys cost ###
m.usd.spend.opt()

# data
m.usd.spend.show()
#######################################################################################################
```

### Q3: Minimize Emissions

```python
### ### A3 ### ###
### min co2 emissions ###
m.gwp.emit.opt() # lowest total CO2 emissions for the system

# data
print('emissions constraints:')
m.emit.show()
print('')
print('the total possible emissions of each CO2 producing technology based on their capacities:')
print('natural gas annual co2 emissions:', co2_ng,'kg/kWy')
print('biomass annual co2 emissions:', co2_bm,'kg/kWy')
print('grid electricity co2 emissions:', co2_elec,'kg/kWy')
print('capacities of the processes running in our model:')
m.emit.output() # based on CAPCITIES
#######################################################################################################
```

### Q3: Maximize Efficiency

```python
### ### A3 ### ###
### max efficiency ###
print('maximizing efficiency:') # in kW
m._.lb(sum(m._.consume)) # takes in all the capacities as a sum and then optimizes those together

print('potential capacities based on operating horizon (in kW):'), m.capacity.output()
#######################################################################################################
```
