Using the simple examples of sorting used vs fresh paper from a mixed pile of both.

Added a new resources of clean paper from the printer.

```python
# from energia import * ## for when you don't know what to import
from energia import Model, Resource, Process
```

```python
# paper_mixed = 20
# paper_dirty = 15
# paper_clean = 5

m = Model()
m.paper_mix = Resource(label = 'Mixed Paper')
m.paper_clean = Resource(label = 'Clean Paper')
m.paper_dirty = Resource(label = 'Dirty Paper')
m.mins = Resource(label = 'Time Taken')
# new resource
m.paper_printer = Resource(label='Printer Paper')
m.paper_printer.consume = True
m.resources
# -----

# sorting
m.sorting = Process()
m.sorting(m.paper_clean) == -4*m.paper_mix + 3*m.paper_dirty - m.mins
m.sorting.operate == True
m.sorting.conversion

# new process
m.go_get_printer_paper = Process()
m.go_get_printer_paper(m.paper_clean) == -1*m.paper_printer - 1.5*m.mins
# takes 1.5 min to go get paper, can change this
m.go_get_printer_paper.operate == True
# -----
m.paper_clean.release >= 20
m.paper_mix.consume == True
m.paper_dirty.release == True
m.mins.consume == True
# -----
# locate processes
m.locate(m.sorting)
# locate new process
m.locate(m.go_get_printer_paper)
# -----
# optimizing objectives
#m.mins.consume.opt()
#m.paper_printer.consume.opt() # it's reading paper_printer as a boolean
m.sorting.operate.opt()
m.sol()
```

```python
# them run this to show the inputs you wrote (indices, bound constraints, calculation constraints, general resource balance constraints)
# aka the "canonical form"
m.show(True)
```
